### **Les types d'erreurs**

Malgré tout le soin que vous porterez à programmer, des erreurs se glisseront dans vos programmes.

Il en existe trois types principaux :

- `Erreur de syntaxe`:
    Dans un programme, la syntaxe doit être parfaitement correcte : la moindre erreur dans le nom des fonctions ou d'indentation provoquera un arrêt de fonctionnement ou une exécution erratique. 
- `Erreur sémantique`:
    Le programme fonctionne, cependant, vous n'obtenez pas le résultat souhaité. Dans ce cas, c'est que le séquencement des instructions de votre programme n'est pas correct. 
- `Erreur d'exécution`:
    Un élément extérieur vient perturber le fonctionnement normal de votre programme. Ces erreurs, également appelées exceptions, sont dues à des circonstances particulières comme par exemple votre programme doit lire un fichier mais il est absent ou l'utilisateur n'a pas entré la valeur attendue.

#### **Importance de l’indentation en Python**

**Qu’est-ce que l’indentation ?**

- L'**Indentation** fait référence aux espaces au début d’une ligne de code.

- En Python, l’indentation est utilisée pour définir la structure et la hiérarchie des blocs de code.

**Pourquoi l’indentation est-elle importante en Python ?**

- `Structure et lisibilité` : L’indentation rend le code visuelment structuré et facile à suivre.
- `Exigences de syntaxe`: Syntax Requirement: Contrairement à de nombreux autres langages de programmation, l’indentation en Python n’est pas seulement pour la lisibilité, elle est nécessaire pour que le code s’exécute.
- `Définition des blocs de code` : Python utilise l'indentation (généralement quatre espaces) pour définir des blocs de code, tels que ceux à l'intérieur des boucles, des conditions, des fonctions et des classes.

**Règles d’indentation en Python**

- `Cohérence `: Python s’attend à une indentation cohérente. Habituellement, la convention est de 4 espaces par niveau d’indentation.
- `Pas de tabulations et d'espaces mixtes` : L'utilisation à la fois de tabulations et d'espaces peut provoquer des erreurs. Tenez-vous-en à une méthode (généralement des espaces).
- `Erreurs dues à un mauvais alignement:`
    - Une indentation mal alignée provoque IndentationError.
    - Un code indenté à tort peut modifier la logique ou provoquer une erreur de syntaxe.

```python
# Exemple Correcte
for i in range(5):
    print(i)  # Cette ligne est indentée pour faire partie de la boucle
print("Fin")  # print("Fin") # Cette ligne n'est pas indentée et s'exécute après la boucle
```

```python
# Exemple Incorrecte
for i in range(5):
print(i)  # L'indentation manquante provoque une erreur d'indentation
```

**Principaux points à retenir**

- L’indentation est cruciale en Python, à la fois pour la fonctionnalité et la lisibilité.
- Indentez systématiquement votre code pour éviter les erreurs.
- N’oubliez pas : en Python, l’indentation est la seule façon de définir des blocs de code.

### **Les nombres entiers de type `int`**

Les `int` représentent le type le plus facilement représentable sur une architecture donnée. 

Par exemple, sur une machine `32-bits`, la taille d'un `int` sera de `32 bits`, donc un `int` permettra de représenter des nombres entre -2^31 et 2^31 - 1, soit entre -2.147.483.648 et 2.147.483.647.

Un littéral `int` s'écrit tout simplement avec les chiffres de 0 à 9, précédé éventuellement du symbole `-`. Il est possible d'écrire ce littéral suivant trois bases :

- **Base décimale** : le littéral devra commencer par un chiffre entre 1 et 9.
- **Base octale (base 8)** : le littéral devra commencer par `0` suivi de chiffres de 0 à 7.
- **Base hexadécimale (base 16)** : le littéral devra commencer par `0x` suivi de chiffres de 0 à 9 et de lettres de A à F (en minuscule ou majuscule).


```python
# Exemples de Quelques entiers int
x=1
x=0
x=-33
x=4566
x=2147483647
x=0x76 #équivalent à x=62
x=0xFF #équivalent à x=255
x=0xa1 #équivalent à x=161
```

### **Les nombres entiers de type long**

Un entier long est un entier dont la taille n'est limitée que par la mémoire allouée par l'ordinateur à l'interpréteur Python. C'est à dire qu'un long permet d'écrire des entiers aussi grands que l'on veut.

Il existe deux manières d'utiliser des long :

- il faut rajouter L ou l à la fin d'un littéral entier pour qu'il soit automatiquement long
- lorsque le résultat d'une opération dépasse la capacité de stockage d'un int, alors, ce résultat est automatiquement convertit en long

```python
# Exemples des Quelques entiers long
#x=1L --> Pas de long en Python 3
#x=-45l --> Pas de long en Python 3
x=121212121212121212121212121 # Automatiquement converti en long
x=2147483647+1
```

Ce type long permet l'encodage de valeurs entières avec une précision quasi infinie : une valeur définie sous cette forme peut en effet posséder un nombre de chiffres significatifs quelconque, ce nombre *n'étant limité que par la taille de la mémoire disponible sur l'ordinateur utilisé* !

La progression continue avec des nombres de plus en plus gigantesques, mais la vitesse de calcul diminue. Les nombres mémorisés sous le type long occupent une place variable dans la mémoire de l'ordinateur, en fonction de leur taille. 

### **Les nombres à virgule flottante (`float`)**

Un nombre à virgule flottante est un nombre décimal qu'il est possible de représenter par sa mantisse et son exposant. Par exemple, le nombre 125,789 est représentable par le couple (mantisse = 1,25789, exposant = 2). La mantisse étant toujours comprise entre -10 et 10 exclus.

Les nombres sont traduits par la formule `nombre = matrisse ∗ 10^{exposant}`.

Les limites dépendent de l'architecture de la machine et sont équivalentes au type de donnée double du langage C.

Les littéraux peuvent s'écrire avec les chiffres, le caractère point pour indiquer la séparation entre la partie entière et la partie décimale et la lettre 'e' ou 'E' pour spécifier l'exposant. 

**Float** (`float`): Nombres décimaux.

**Use case**: Calculs scientifiques, mesures, données financières.

```python
# Exemples des nombres à virgules
x = 1.234
x = 1.0 # Notons qu'un entier peut être un flottant
x = 1. # Même résultat que précédemment
x = 1.234e54 # C'est à dire <math>1.234*10^{54}</math>
x = 1.234E54 # idem
x = -1.454e-2 #La mantisse et l'exposant peuvent être négatifs
```

### **Les nombres complexes**

Python est un des rares langages à proposer un type de base pour les nombres complexes `*C*`. Un nombre complexe est un nombre composé d'une partie réelle et d'une partie imaginaire. On note `*i*` une des racines du polynôme `x^2 = − 1`.

Dans certains contextes, (comme la physique), la racine de -1 est notée `j`. C'est ce qui se fait en Python.

Un littéral complexe s'écrit donc : `a` + `bj`, avec `a` et `b` des variables de type float. 

Attention, `j` doit être précédé d'un nombre car sinon, Python l'interprétera comme étant la variable `j`. Il ne doit pas `y` avoir d'espace entre ce nombre et `j`.

```python
foin = "Cette leçon vaut bien un fromage, sans doute ?"
aiguille = "fromage"
print(foin.find(aiguille))      # Trouve l'index de la première occurrence de l'aiguille
```

- `rfind(aiguille)` : pareil que `"find"` mais en partant de la fin (reverse).

```python
foin = "Cette leçon vaut bien deux fromages, dont un fromage râpé ?"
aiguille = "fromage"
print(foin.rfind(aiguille))     # Recherche de la dernière occurrence de l'aiguille
```

- `count(aiguille)` : compte le nombre de sous-chaînes aiguille dans la chaîne.

```python
foin = "Le héron au long bec emmanché d'un long cou"
aiguille = 'long'
print(foin.count(aiguille))     # foin.count('long') renvoie 2  
```

- `lower()` : convertit une chaîne en minuscules.

```python
phrase ="ATTENTION : Danger !"
print(phrase.lower())       # Convertit la chaîne en minuscules
```

- `upper()` : convertit une chaîne en majuscules.

```python
phrase = "Merci beaucoup"
print(phrase.upper())       # Output: MERCI BEAUCOUP
```

- `capitalize()` : convertit en majuscule la première lettre d'une chaîne.

```python
phrase = "quel beau temps, aujourd'hui !"
print(phrase.capitalize())      # Mettre la première lettre en majuscule
```

- `swapcase()` : convertit toutes les majuscules en minuscules et vice-versa.

```python
phrase = "La CIGALE et la FOURMI"
print(phrase.swapcase())  # Inverse la casse de chaque lettre
```

- `strip()` : enlève les espaces éventuels au début et à la fin de la chaîne (trime).

```python
phrase = "   Monty Python   "
phrase.strip()
```

- `replace(old, new)` : remplace tous les caractères old par des caractères new dans la chaîne.

```python
phrase = "Si ce n'est toi c'est donc ton frère"
print(phrase.replace(" ","_"))
```

 Dans la plupart de ces méthodes, il est possible de préciser quelle portion de la chaîne doit être traitée, en ajoutant des arguments supplémentaires. 

```python
# Initialisation de la variable ch9
ch9 = "exemple de chaîne"

# Recherche de l'indice du premier caractère "e"
print(ch9.index("e"))  # Retourne l'indice de la première occurrence de "e"
```

```python
print(ch9.index("e",5))		# cherche seulement à partir de l'indice 5
							# et trouve le second 'e'
```

```python
print(ch9.index("e",15))		# cherche à partir du caractère n° 15
							    # et trouve le quatrième 'e'
```
