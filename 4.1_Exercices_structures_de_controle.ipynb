1. Écrivez un programme qui convertisse en radians un angle fourni au départ en degrés, minutes, secondes.

```python
import math
# 1. Saisie des données par l'utilisateur
deg = float(input("Entrez les degrés : "))
min = float(input("Entrez les minutes : "))
sec = float(input("Entrez les secondes : "))
# 2. Conversion DMS (Degrés, Minutes, Secondes) en degrés décimaux
deg_decimaux = deg + (min / 60) + (sec / 3600)

# 3. Conversion en radians (degrés * pi / 180)
radians = deg_decimaux * (math.pi / 180)

# 4. Affichage du résultat
print(f"{deg}° {min}' {sec}\" correspond à {radians:.5f} radians.")
```

2. Écrivez un programme qui convertisse en degrés, minutes, secondes un angle fourni au départ en radians.

```python
radians = float(input("Entrez l'angle en radians à convertir en DMS : "))

# 1. Conversion des radians en degrés décimaux
degres_decimaux = math.degrees(radians)
    
# 2. Extraire la partie entière pour les degrés
degres = int(degres_decimaux)
    
# 3. Calculer les minutes à partir du reste décimal
reste_minutes = (degres_decimaux - degres) * 60
minutes = int(reste_minutes)
    
# 4. Calculer les secondes à partir du reste des minutes
secondes = (reste_minutes - minutes) * 60
    

print(f"L'angle est de : {degres}° {minutes}' {secondes:.2f}''")
```

3. Écrivez un programme qui convertisse en degrés Celsius une température exprimée au départ en degrés Fahrenheit, ou l'inverse. 
La formule de conversion est : `Tf = Tc * 1.8 + 32`

```python

print("--- Convertisseur de Température ---")
choix = input("Voulez-vous convertir vers (C)elsius ou (F)ahrenheit ? ").upper()

if choix == 'F':
    temp = float(input("Entrez la température en Celsius : "))
    resultat = temp * 1.8 + 32
    print(f"{temp}°C correspondent à {resultat:.2f}°F")
elif choix == 'C':
    temp = float(input("Entrez la température en Fahrenheit : "))
    resultat = (temp - 32)/1.8
    print(f"{temp}°F correspondent à {resultat:.2f}°C")
else:
    print("Choix invalide. Veuillez entrer 'C' ou 'F'.")
```

5. Écrivez un programme qui calcule les intérêts accumulés chaque année pendant 30 ans, par capitalisation d'une somme de 100 euros placée en banque au taux fixe de 3,5 %

```python
capital = 100.0  # Somme de départ
taux = 0.035     # Taux fixe de 3,5 %
annees = 30

print(f"{'Année':<10} | {'Capital accumulé':<20}")
print("-" * 35)

for i in range(1, annees + 1):
    # Calcul des intérêts capitalisés : Capital * (1 + taux)
    capital = capital * (1 + taux)
    print(f"Année {i:<4} | {capital:>15.2f} €")

print("-" * 35)
print(f"Total après 30 ans : {capital:.2f} €")
```

7. Analyse des nombres Demandez à l'utilisateur une série de nombres séparés par des espaces. Effectuez les tâches suivantes :

- Convertissez les nombres en une liste d'entiers.
- Affichez le plus grand et le plus petit nombre.
- Affichez la somme de tous les nombres.
- Affichez les nombres pairs.

    Exemple attendu :

        `Entrez des nombres séparés par des espaces : 3 5 8 2 7`
        `Le plus grand : 8`
        `Le plus petit : 2`
        `La somme : 25`
        `Nombres pairs : [8, 2]`

```python
# 1. Demander la saisie à l'utilisateur
entree = input("Entrez des nombres séparés par des espaces : ")

# 2. Convertir la saisie en une liste d'entiers
# On utilise une compréhension de liste pour transformer chaque chaîne en int
nombres = [int(n) for n in entree.split()]

# 3. Effectuer les analyses
plus_grand = max(nombres)
plus_petit = min(nombres)
somme_totale = sum(nombres)

# 4. Filtrer les nombres pairs avec modulo % 2
pairs = [n for n in nombres if n % 2 == 0]

# Affichage des résultats
print(f"Le plus grand : {plus_grand}")
print(f"Le plus petit : {plus_petit}")
print(f"La somme : {somme_totale}")
print(f"Nombres pairs : {pairs}")
```

8. Créez un jeu où le programme choisit un nombre aléatoire entre 1 et 100, et l'utilisateur doit deviner le nombre. Après chaque tentative, affichez un message indiquant si le nombre est "plus grand" ou "plus petit". Arrêtez le jeu lorsque l'utilisateur devine correctement.

    Exemple attendu :

        `Devinez le nombre : 50`
        `Trop grand !`
        `Devinez le nombre : 30`
        `Trop petit !`
        `Devinez le nombre : 40`
        `Bravo, vous avez trouvé le nombre !`

```python

import random

# Générer un nombre aléatoire entre 1 et 100
nombre_secret = random.randint(1, 100)
tentative = 0

print("--- Jeu du Juste Prix (1 à 100) ---")

while True:
    # Saisie du joueur
    essai = int(input("Devinez le nombre : "))
    tentative += 1
    
    if essai > nombre_secret:
        print("Trop grand !")
    elif essai < nombre_secret:
        print("Trop petit !")
    else:
        print(f"Bravo, vous avez trouvé le nombre en {tentative} tentatives !")
        break
```

10. Créez un jeu où le programme génère un nombre aléatoire entre 1 et 100 et demande à l'utilisateur de deviner ce nombre. Ajoutez les fonctionnalités suivantes :

    - Limitez le nombre de tentatives à 7.
    - Donnez des indices après chaque tentative (par exemple, "plus grand" ou "plus petit").
    - Affichez le score final basé sur le nombre de tentatives restantes.

```python
import random

def jouer_jeu():
    nombre_secret = random.randint(1, 100)
    tentatives_max = 7
    gagne = False

    print(f"--- Jeu du Juste Prix : 1 à 100 (Max {tentatives_max} essais) ---")

    for essai in range(1, tentatives_max + 1):
        devinette = int(input(f"Tentative {essai}/{tentatives_max} - Entrez un nombre : "))

        if devinette == nombre_secret:
            # Score basé sur les tentatives restantes
            score = (tentatives_max - essai) + 1
            print(f"Bravo, vous avez trouvé le nombre ! Score final : {score}/7")
            gagne = True
            break
        elif devinette < nombre_secret:
            print("Plus grand !")
        else:
            print("Plus petit !")

    if not gagne:
        print(f"Perdu ! Le nombre était {nombre_secret}. Score final : 0")

jouer_jeu()
```

22 Créez un programme qui demande à l'utilisateur de saisir les valeurs du tableau.
Lorsque l'utilisateur saisi 'q' ou 'Fin', le programma affiche les contenus du tableau
et calcule la moyenne

```python
# 1. Demander la saisie à l'utilisateur

tableau = []
while True:
   entree = input("Entrez des nombres  : ").strip()
   if entree == "q" or entree == "Fin":
       break
   tableau.append(float(entree))
  
   
print(f"Voici le tableau final : {tableau}")

if len(tableau) == 0:
    print("Le tableau est vide. Impossible de calculer la moyenne.")
    exit()
else:
    print(f"La moyenne des éléments du tableau est : {sum(tableau)/len(tableau)}")
    
```
